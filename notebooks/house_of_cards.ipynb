```python
import os
import sys
import numpy as np
import copy as cp
import math
import matplotlib.pyplot as plt
import matplotlib as mpl
import scipy.stats
import pickle
from datetime import date
```

```python
sys.path.append(os.path.abspath('../code/'))

from HOC_approximate_functions import simulate_approximate_HOC
from HOC_functions import simulate_hoc_wrapper
```

```python
# whether to save pdfs
save = True

outputdir = os.path.join ('../results/HoC', str (date.today ()))
os.makedirs(outputdir, exist_ok=True)

# colors for different scenarios
scenarios  = ['adaptive','random','beneficial','conditional','deleterious']
mycolors   = ['salmon', 'lightgray', 'gold', 'dimgray','cornflowerblue']
color_dict = dict (zip (scenarios, mycolors))
```

```python
# parameters
nsims  = 100  # number of simulations
L      = 100  # total number of loci
M      = 4    # number of generations to evolve
sigma  = 1    # variance of fitness distribution
ntypes = len (scenarios)    # number of scenarios: adaptive, random, beneficial, conditioned

# parameters for approximate scenario
mean     = 4
variance = 4

# different library sizes
Ms       = [2,4,6]
```

```python
seed = 3151

my_rng = np.random.default_rng (seed)
```

```python
# approximate HoC
R2_list = list ()
Am_list = list ()
Coef_list = list ()
Int_list = list ()

for m in Ms :
    R2_approx, Am_approx, Coef, intercepts, approx_scenarios = simulate_approximate_HOC (m, mean=mean,
                                                                                         variance=variance, nsims=nsims, seed=seed+m)
    R2_list.append (cp.deepcopy (R2_approx))
    Am_list.append (cp.deepcopy (Am_approx))
    Coef_list.append ( cp.deepcopy (Coef))
    Int_list.append ( cp.deepcopy (intercepts))
```

```python
R2_real_list = list ()
Am_real_list = list ()
Max_list = list ()
Path_list = list ()
Traj_list = list ()

for m in Ms :
    
    # Simulate
    R2mat_m, Akmat_m, nmax_step_m, npaths_m, all_dict_m, mylabels_m = simulate_hoc_wrapper ( m, L, nsims, weighted=False, rng=my_rng) 
    
    with open( os.path.join (outputdir, 'R2_' + str (m) + '.pkl'), 'wb') as file:
        pickle.dump(R2mat_m, file)
    with open( os.path.join (outputdir, 'nmax_' + str (m) + '.pkl'), 'wb') as file:
        pickle.dump(nmax_step_m, file)
    np.savetxt ( os.path.join (outputdir, 'npaths_' + str (m) + '.txt'), npaths_m )
    
    R2_real_list.append (cp.deepcopy (R2mat_m))
    Am_real_list.append (cp.deepcopy (Akmat_m))
    Max_list.append (cp.deepcopy (nmax_step_m))
    Path_list.append (cp.deepcopy (npaths_m))

    Ys = np.zeros ((nsims, m+1))
    for i in range (nsims) :
        Ys[i,:] = cp.deepcopy (all_dict_m['adaptive'][i][2])
    
    Traj_list.append (cp.deepcopy (Ys))
```

```python
for i in range (len(Traj_list)) :
    np.savetxt (os.path.join (outputdir, 'trajectory_' + str (Ms[i]) + '.txt'), Traj_list[i])
```
